```python
import json
import pandas as pd

etiquetas = tuple(
    ["real"] * 250 +
    ["deepfakes"] * 150 +
    ["face2face"] * 150 +
    ["faceshifter"] * 150 +
    ["faceswap"] * 150 +
    ["neuraltextures"] * 150
)

# 2. Leemos el archivo Excel
df = pd.read_excel(
    r"D:\OneDrive - UNIR\Master\99.TFM\DataSets\Resultados_Analisis\DeepFakeDetector.xlsx"
)

# 3. Construimos la estructura del diccionario/JSON
resultado = {}

for index, row in df.iterrows():
    posicion_excel = int(row["Posicion"])

    # Dado que las posiciones en Excel van de 1 a 1000 y en Python indexan desde 0,
    # la posición 1000 corresponde al índice 999 de la tupla:
    indice_tupla = posicion_excel - 1

    # Obtenemos el nombre de la clasificación de la tupla
    clasificacion = etiquetas[indice_tupla]

    veredicto = row["Verdict"]
    ai_score = float(row["AI Score (%)"])

    # Generamos la clave identificativa (Imagen1, Imagen2, etc., basándonos en el número de fila/índice)
    id_imagen = f"Imagen{index + 1}"

    # Añadimos la entrada con la estructura requerida
    resultado[id_imagen] = {
        "clasificacion": clasificacion,
        "Veredicto": veredicto,
        "AI Score": ai_score,
    }

# 4. Guardamos el resultado en un archivo JSON
with open("DeepFakeDetector_clean.json", "w", encoding="utf-8") as file:
    json.dump(resultado, file, ensure_ascii=False, indent=4)

print("¡Archivo JSON generado con éxito!")
```

```python
import json

# 1. Creamos la tupla de etiquetas reducida (100 elementos en total)
etiquetas_hive = tuple(
    ["real"] * 25
    + ["deepfakes"] * 15
    + ["face2face"] * 15
    + ["faceshifter"] * 15
    + ["faceswap"] * 15
    + ["neuraltextures"] * 15
)

# 2. Cargar el JSON de respuestas de Hive
# Cambia 'RespuestasHive.json' por la ruta correcta de tu archivo
with open(r"D:\OneDrive - UNIR\Master\99.TFM\DataSets\Resultados_Analisis\respuestasHive.json", "r", encoding="utf-8") as file:
    datos_hive = json.load(file)

resultado = {}

# 3. Procesamos cada entrada en el JSON
for index, item in enumerate(datos_hive):
    # Asignar la clasificación según el índice actual en la tupla
    clasificacion = etiquetas_hive[index]

    # Extraer el array de clases de la respuesta de Hive
    classes = item["output"][0]["classes"]

    # Buscar los valores de "not_ai_generated" y "ai_generated"
    not_ai_val = next(
        (c["value"] for c in classes if c["class"] == "not_ai_generated"), None
    )
    ai_generated_val = next(
        (c["value"] for c in classes if c["class"] == "ai_generated"), None
    )

    # Determinar veredicto y AI Score en base a la probabilidad de ai_generated
    veredicto = "AI Generated" if ai_generated_val > not_ai_val else "Human"
    ai_score = round(ai_generated_val * 100, 2)  # Convertir a porcentaje

    # Construir la entrada del diccionario
    id_imagen = f"Imagen{index + 1}"
    resultado[id_imagen] = {
        "clasificacion": clasificacion,
        "Veredicto": veredicto,
        "AI Score": ai_score,
    }

# 4. Guardar el nuevo JSON
with open("Hive_clean.json", "w", encoding="utf-8") as file:
    json.dump(resultado, file, ensure_ascii=False, indent=4)

print("¡JSON procesado y guardado correctamente!")
```

```python
import json

# 1. Creamos la tupla de etiquetas de 1000 elementos
etiquetas = tuple(
    ["real"] * 250
    + ["deepfakes"] * 150
    + ["face2face"] * 150
    + ["faceshifter"] * 150
    + ["faceswap"] * 150
    + ["neuraltextures"] * 150
)

# 2. Cargar el JSON original que contiene los 1000 elementos
# Reemplaza 'respuestas_originales.json' por la ruta de tu archivo
with open(r"D:\OneDrive - UNIR\Master\99.TFM\DataSets\Resultados_Analisis\Sightengine.json", "r", encoding="utf-8") as file:
    datos_api = json.load(file)

resultado = {}

# 3. Iterar sobre la lista de objetos
for index, item in enumerate(datos_api):
    # Obtenemos la clasificación usando el índice del bucle en la tupla de 1000 elementos
    clasificacion = etiquetas[index]

    # Extraer el valor de deepfake desde respuesta_api -> type -> deepfake
    deepfake_score = item.get("respuesta_api", {}).get("type", {}).get("deepfake", 0.0)

    # Convertir a porcentaje
    ai_score = round(deepfake_score * 100, 2)

    # Determinar el veredicto
    veredicto = "AI Generated" if ai_score >= 50.0 else "Human"

    # Generar la clave identificativa
    id_imagen = f"Imagen{index + 1}"

    # Construir la entrada final
    resultado[id_imagen] = {
        "clasificacion": clasificacion,
        "Veredicto": veredicto,
        "AI Score": ai_score,
    }

# 4. Guardar el resultado procesado
with open("Sightengine_clean.json", "w", encoding="utf-8") as file:
    json.dump(resultado, file, ensure_ascii=False, indent=4)

print("¡Archivo JSON generado con éxito usando la tupla de 1000 etiquetas!")
```
